# Filter Isopleth Data by Solar Radiation

Join isopleth ozone data with solar radiation observations and remove rows below a configurable `SR_THRESHOLD`.

```python
from pathlib import Path

import pandas as pd

ISOPLETH_PATH = Path("/Users/isaacsudweeks/Desktop/Work/Improved_Isopleths/data/utah/data_up_to_2025/Speciated/HW_data_raw_2026-05-23_winter.csv")
SR_PATH = Path("../data/SR_Data/SR_Data_HW.csv")
SR_THRESHOLD = 710  # W/m^2; adjust as needed

```

```python
isopleth_df = pd.read_csv(ISOPLETH_PATH, parse_dates=["dt"]) #Change dt to window_start for canister data
# isopleth_df['dt'] = isopleth_df['window_start']

sr_df = (
    pd.read_csv(SR_PATH, index_col=0, parse_dates=True)
    .rename_axis("dt")
    .reset_index()
)

isopleth_df.head()
```

```python
merged_df = isopleth_df.merge(
    sr_df,
    on="dt",
    how="inner",
    # validate="one_to_one",
)

filtered_df = merged_df[merged_df["solar_radiation_w_m2"].ge(SR_THRESHOLD)].copy()
filtered_df.head()
```

```python
print(
    f"Rows retained: {len(filtered_df)} of {len(merged_df)} (threshold >= {SR_THRESHOLD} W/m^2)"
)
filtered_df.tail()
```

```python
filtered_df.to_csv("/Users/isaacsudweeks/Desktop/Work/Improved_Isopleths/data/utah/data_up_to_2025/Speciated/SR_filtered/summer_2025_speciated.csv",index=False)
```
